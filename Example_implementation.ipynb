## Prediction of linear site terms using microtremor horizontal-to-vertical spectral ratios (mHVSR)

The notebook is intended to illustrate the prediction of site terms using mHVSR based on Artificial Neural Network (ANN) Ensemble models. 

```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
import tensorflow as tf
from tensorflow.keras.layers import Average
import os
```

```python
#Load mHVSR data

hvsr_df = pd.read_csv('HVSR_data.csv')
display(hvsr_df.head())
```

```python
model_dir = r'ANN1/'
site_term_freqs = np.sort([float(i) for i in os.listdir(model_dir)])
```

```python
#Predicting site terms 

input_vals = np.log(hvsr_df.iloc[:,3:])
predictions_df = hvsr_df.iloc[:,:3]

for f in site_term_freqs:
    models = [tf.keras.models.load_model(fr'ANN1/{f}/model_{k}.keras') for k in range(1,8)]  
    predictions = np.array([m.predict(input_vals,verbose = 0).ravel() for m in models])
    ensemble_predictions = np.mean(predictions,axis=0)
    predictions_df[f'SiteTerm@{f}Hz']  = ensemble_predictions 
```
